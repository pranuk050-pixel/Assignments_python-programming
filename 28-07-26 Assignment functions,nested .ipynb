```python
#Assignment 13: Factorial Using a Function
#Write a function named calculate_factorial().
#Accept a non-negative integer.
#Calculate its factorial using a loop.
#Do not use math.factorial().
#Return the factorial.
#Handle negative input.
#Example Input :Enter a number: 5  Expected Output Factorial: 120

def calculate_factorial(number):   #function name and parameter
    factorial = 1
    if number < 0:      # error msg if less than 0
        return "Factorial is not defined for negative numbers."
    elif number == 0:
        return 1
    else:
        for i in range(1, number + 1):   #start factorial from 1 till 5
            factorial *= i
        return factorial

number = int(input("Enter a number: "))
factorial_result = calculate_factorial(number)
print(f"Factorial: {factorial_result}")

```

```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Requirements
# Accept a list of numbers.
# Do not use max().
# Return the largest list element.
# Handle an empty list by returning an appropriate message.
def find_list_maximum(numbers):
  if not numbers:                 # checking empty list
    return "Empty list"
  maximum = numbers[0]    # assuming 1st element is largest
  for num in numbers:     #looping through list
    if num > maximum:
      maximum = num
  return maximum

numbers = input("Enter numbers: ").split()   #user input
numbers = [int(num) for num in numbers]      # conversion string to integer
maximum = find_list_maximum(numbers)
print("Maximum:", maximum)

```

```python
# Assignment 32: Function Returning a Dictionary
# Write a function named analyze_text().

# Requirements
# Accept a string.
# Return a dictionary containing:
# Total characters
# Total words
# Total vowels
# Total digits
# Total spaces
# Display every result clearly.

def analyze_text(text):
  char_count = len(text)
  word_count = len(text.split())
  vowel_count = 0     #starting  count with 0
  digit_count = 0
  space_count = 0
  vowels = "aeiouAEIOU"
  for char in text:
    if char in vowels:   #if in vowels count and move to next vowel count
      vowel_count += 1
    elif char.isdigit(): #if its digit then count and move to count next digit
      digit_count += 1
    elif char.isspace():  #count of spaces
      space_count += 1
    else:
      pass
  result = {
    "Total characters": char_count,
    "Total words": word_count,
    "Total vowels": vowel_count,
    "Total digits": digit_count,
    "Total spaces": space_count}
  return result
text = input("Enter a text: ")
result = analyze_text(text)
for key, value in result.items():
  print(f"{key}: {value}")

```

```python
# Assignment 35: Fibonacci Function
# Write a function named generate_fibonacci().

# Requirements
# Accept the number of terms.
# Return the complete Fibonacci series as a list.
# Handle zero and negative term counts.
# Do not use recursion in this assignment.
# Example Input
# Enter number of terms: 7
# Expected Output
# [0, 1, 1, 2, 3, 5, 8]

def generate_fibonacci(n):
  if n <= 0:      #ensuring no negative terms
    return []
  elif n == 1:
    return [0]
  elif n == 2:
    return [0, 1]
  else:
    fibonacci_series = [0, 1]   #start series  with 1st two terms
    for i in range(2, n):
      next_term = fibonacci_series[-1] + fibonacci_series[-2] #next term =sum of last two terms
      fibonacci_series.append(next_term)   #appending next term to the series
  return fibonacci_series
terms=int(input("Enter number of terms: "))
fibonacci_series = generate_fibonacci(terms)
print(fibonacci_series)

```

```python
# Assignment 91: Closure-Based Counter
# Write a function named create_counter().

# Requirements
# Define a variable count = 0 in the outer function.
# Define an inner function that increases the count.
# Use nonlocal to modify the enclosing variable.
# Return the inner function.
# Call the returned function at least five times.
# Expected Output
# 1
# 2
# 3
# 4
# 5

def create_counter():
  count = 0          #variable count
  def increment():
    nonlocal count  #used non local to modify the enclosing variable
    count += 1
    return count
  return increment  #add to count value
counter = create_counter()
print(counter())
print(counter())
print(counter())
print(counter())
print(counter())
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# Requirements
# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry
# Use nonlocal to update the balance.
# Reject withdrawals greater than the available balance.
# Return all three inner functions.
# Hint
# Return multiple functions:

# return deposit, withdraw, check_balance

def create_bank_account():
  balance = 0
  def deposit(amount):
    nonlocal balance
    balance += amount  #adding to the balance
    return balance   #new balance after deposit
  def withdraw(amount):
    nonlocal balance
    if amount > balance:
      return "insufficient balance"
    balance -= amount   #duduction from balance
    return balance
  def check_balance():   #checking balance after withdrawal
    return balance
  return deposit,withdraw,check_balance

deposite,withdraw,check_balance =create_bank_account()
print(deposite(5200))
print(check_balance())
print(withdraw(2000))
print(check_balance())
print(withdraw(1000))
print(check_balance())
print(withdraw(3000))
print(check_balance())
```

```python
# Assignment 93: Function Decorator
# Create a decorator named execution_logger().

# Requirements
# The decorator should print a message before the original function executes.
# It should print another message after execution.
# Apply it to an addition function.
# Preserve the original function's return value.
# Expected Pattern
# Starting function...
# Result: 30
# Function completed.

def execution_logger(func):
  def wrapper(*args,**kwargs):
    print("starting function...")
    result = func(*args,**kwargs)
    print("result:",result)
    print("function completed.")
    return result
  return wrapper

@execution_logger   #using decorator
def addition(a,b):
  return a + b
result = addition(10,20)

```
