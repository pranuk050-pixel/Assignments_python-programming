### Q1. Find a account code for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

# a line of text like what a teller or app log might show
text = "Card ending statement for ACC-1001: withdrawal processed at counter 4"

# ACC- followed by exactly 4 digits, with word boundaries so we don't
# accidentally grab part of a longer code
pattern = re.compile(r"\bACC-\d{4}\b")

match = pattern.search(text)
print("account code found:", match.group() if match else "none")

# edge case: nothing to find
print("empty text check:", pattern.search("") is None)
```

**Explanation for Q1:**

**How this works:** `\bACC-\d{4}\b` looks for the literal text `ACC-` followed by
exactly four digits, with `\b` (word boundary) on both ends so it won't match something like
`XACC-1000` or `ACC-10001`. `.search()` is enough here since we just want to know if the code
shows up anywhere in the sentence. On an empty string it correctly returns `None`, which is
exactly the "not found" edge case we want to guard against before using the result.

### Q2. Validate a six-digit pin for banking case 1; use sample reference `ACC-1002` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

# (ACC-1002) sample PINs a customer might type at an ATM
pin_candidates = ["482913", "12345", "ab1234", "0000000", "930215"]

# exactly six digits, nothing more, nothing less
pin_pattern = re.compile(r"^\d{6}$")

for pin in pin_candidates:
    print(pin, "->", "valid" if pin_pattern.fullmatch(pin) else "invalid")
```

**Explanation for Q2:**

**How this works:** `fullmatch` (or `^...$`) is the right tool for validation, because it
forces the *whole* string to satisfy the pattern instead of just a piece of it. `\d{6}` alone
would happily match the first six digits of "0000000" and call it valid — that's why we anchor
it. A 5-digit PIN, a 7-digit PIN, and letters mixed with digits are all correctly rejected.

```python
import re

statement = "ACC-1003 charges: INR 1211.25 processing fee, INR 21.5 SMS charge, refund INR 0.99"

# whole or decimal numbers (up to 2 decimal places)
amounts = re.findall(r"\d+(?:\.\d{1,2})?", statement)
print("amounts found:", amounts)
```

decimal point plus one or two digits — the `?` makes the whole decimal part optional so plain
whole numbers still match. `findall` then returns every amount in the sentence, in order,
as plain strings (you'd `float()` them before doing any arithmetic).

### Q8. Check a phone number for banking case 1; use sample reference `ACC-1008` and explain the result.

**Hints**

```python
import re

numbers = ["+91 98765 43210", "9876543210", "98765-43210", "12345", "5876543210"]
phone_pattern = re.compile(r"^(?:\+91)?[6-9]\d{9}$")

for n in numbers:
    cleaned = re.sub(r"[\s-]", "", n)   # strip spaces/dashes before checking
    print(n, "->", bool(phone_pattern.fullmatch(cleaned)))
```

### Q10. Replace masked card digits for banking case 1; use sample reference `ACC-1010` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

card = "4111 1111 1111 2010"

def mask_card(number: str) -> str:
    digits = re.sub(r"\D", "", number)          # keep only the digits
    return re.sub(r"\d(?=\d{4})", "*", digits)  # star out everything except the last 4

print("original:", card)
print("masked:  ", mask_card(card))
```

**Explanation for Q10:**

**How this works:** first strip out spaces so we're working with a plain digit string, then
use a lookahead `(?=\d{4})` to only replace a digit if it's followed by at least four more
digits — that's what protects the last four digits from being masked while everything before
them turns into `*`.

### Q13. Find email addresses for banking case 1; use sample reference `ACC-1013` and explain the result.

```python
import re

support_text = "Query about ACC-1013 sent from raj.kumar99@examplebank.com, cc ops@examplebank.co.in"

emails = re.findall(r"[\w.+-]+@[\w-]+\.[\w.-]+", support_text)
print(emails)
```

by the rest of the domain. It's a practical, "good enough for extraction" pattern — real email
validation is a much deeper rabbit hole, so this favours simplicity over covering every RFC
edge case.

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re
import time

# risky pattern to AVOID: nested quantifiers like (A+)+ can force the engine
# into exponential backtracking on certain non-matching input
# risky_pattern = re.compile(r"^(A+)+B$")

# safe rewrite: a single quantifier does the same job without the nesting
safe_pattern = re.compile(r"^A+B$")

test_input = "A" * 25 + "C"   # deliberately does NOT end in B

start = time.time()
print("matches:", bool(safe_pattern.fullmatch(test_input)))
print("time taken:", round(time.time() - start, 5), "seconds")
```

**Explanation for Q78:**

**How this works:** the danger in patterns like `(A+)+B` is that the engine can split a run
of A's into a huge number of equivalent groupings while trying to satisfy the outer `+`, and each
one gets tried before giving up — that's catastrophic backtracking. Flattening it to a single
`A+` removes the ambiguity entirely, so even a long non-matching string like this one resolves
instantly instead of hanging.
