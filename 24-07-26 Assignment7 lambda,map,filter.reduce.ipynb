Section 5: Combined Lambda, map(), filter(), and reduce() — 20 Questions

```python
#Question 1
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
#Using filter() and map():

#Extract all even numbers.
#Calculate the square of every even number.
#Return the squared values as a list.
even_num = list(filter(lambda x: x%2==0,numbers))
squared_even_num = list(map(lambda x: x**2,even_num))
print("even numbers:",even_num)
print ("squared values:",squared_even_num)
```

```python
#Question 2
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
#Using filter() and reduce():

#Extract all odd numbers.
#Calculate their total sum.

odd_num = list(filter(lambda x: x%2!=0,numbers))
total_sum = reduce(lambda x,y: x+y,odd_num)
print("odd numbers:",odd_num)
print("total sum:",total_sum)
```

```python
#Question 3
numbers = [10, 25, 40, 55, 70, 85, 100]
#Using filter() and map():

#Extract numbers greater than 50.
#Multiply every selected number by 2.

greater_than_50 = list(filter(lambda x: x>50,numbers))
doubled_values = list(map(lambda x: x*2,greater_than_50))
print("numbers greater than 50:",greater_than_50)
print("doubled values:",doubled_values)
```

```python
#Question 4
numbers = [-15, 20, -35, 40, 50, -10, 60]
#Using filter(), map(), and reduce():

#Extract positive numbers.
#Calculate the square of every positive number.
#Calculate the total sum of the squared values.

positive_num = list(filter(lambda x: x>0,numbers))
squared_values = list(map(lambda x: x**2,positive_num))
total_sum = reduce(lambda x,y: x+y, squared_values)

print("positive numbers:",positive_num)
print("squared values of positive numbers:",squared_values)
print("total sum of the squared values:",total_sum)
```

```python
#Question 5
words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]
#Using filter() and map():

#Extract words containing more than five characters.
#Convert the selected words to uppercase.

words_more_than_five = list(filter(lambda x: len(x)>5,words))
upper_case_words = list(map(lambda x: x.upper(),words_more_than_five))

print("words containing more than five characters:",words_more_than_five)
print("upper case words:",upper_case_words)
```

```python
#Question 6
names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]
#Using filter() and map():

#Extract names beginning with "A".
#Convert the selected names to lowercase.

names_beg_A = list(filter(lambda x: x[0]=="A",names))
lower_case_names = list(map(lambda x: x.lower(),names_beg_A))

print("names beginning with A:",names_beg_A)
print("lower case names:",lower_case_names)
```

```python
#Question 7
numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
#Using filter() and reduce():
from functools import reduce
#Extract numbers divisible by both 3 and 5.
#Calculate their total sum.
num_divi_by_3_5 = list(filter(lambda x: x%3==0 and x%5==0,numbers))
total_sum = reduce(lambda x,y: x+y,num_divi_by_3_5)

print("numbers divisible by both 3 and 5:",num_divi_by_3_5)
print("total sum:",total_sum)
```

```python
#Question 8
words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]
#Using filter() and reduce():

#Extract palindrome words.
#Join them into one string separated by " | ".

palindrome_words = list(filter(lambda x: x==x[::-1],words))
join_str = reduce(lambda x,y: x+" | "+y,palindrome_words)

print("palindrome words:",palindrome_words)
print("joined string:",join_str)
```

```python
#Question 9
temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
#Using filter(), map(), and reduce():

#Extract temperatures greater than 30°C.
#Convert the selected values into Fahrenheit.
#Calculate the average Fahrenheit temperature.

greater_than_30 = list(filter(lambda x: x>30,temperatures))
fahrenheit = list(map(lambda x: (x*9/5)+32,greater_than_30))
avg_fahrenheit = reduce(lambda x,y: x+y,fahrenheit)/len(fahrenheit)

print("temperatures greater than 30°C:",greater_than_30)
print("fahrenheit temperatures:",fahrenheit)
print("average fahrenheit temperature:",avg_fahrenheit)
```

```python
#Question 10
prices = [450, 1200, 2500, 800, 5000, 1500, 7500]
#Using filter(), map(), and reduce():

#Extract prices greater than 1000.
#Apply a 10% discount to every selected price.
#Calculate the total discounted amount.

greater_than_1000 = list(filter(lambda x: x>1000,prices))
discount = list(map(lambda x: x*0.9,greater_than_1000))
total_discounted_amount = reduce(lambda x,y: x+y,discount)

print("prices greater than 1000:",greater_than_1000)
print("discounted prices:",discount)
print("total discounted amount:",total_discounted_amount)
```

```python
#Question 11
employees = [
    {"id": 101, "name": "Rahul", "salary": 45000},
    {"id": 102, "name": "Anil", "salary": 65000},
    {"id": 103, "name": "Sneha", "salary": 72000},
    {"id": 104, "name": "Kiran", "salary": 82000},
    {"id": 105, "name": "Meena", "salary": 55000}
]
#Using filter() and map():

#Extract employees earning more than 60000.
#Increase their salary by 10%.
#Return their names and updated salaries.

greater_than_60000 = list(filter(lambda x: x["salary"]>60000,employees))
increase_salary = list(map(lambda x: {"name":x["name"],"salary":x["salary"]*1.1},greater_than_60000))

print("employees earning more than 60000:",greater_than_60000)
print("updated salaries:",increase_salary)
```

```python
#Question 13
employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]
#Using filter(), map(), and reduce():

#Extract active employees.
#Increase every active employee's salary by 15%.
#Calculate the total updated payroll.

active_employees = list(filter(lambda x: x["active"]==True,employees))
increase_salary = list(map(lambda x: {"name":x["name"],"salary":x["salary"]*1.15},active_employees))
total_updated_payroll = reduce(lambda x,y: x+y["salary"],increase_salary,0)

print("active employees:",active_employees)
print("updated salaries:",increase_salary)
print("total updated payroll:",total_updated_payroll)
```

```python
#Question 14
students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]
#Using filter() and reduce():

#Extract students who passed.
#Find the topper among the passed students.

passed_students = list(filter(lambda x: x["passed"]==True,students))
topper = reduce(lambda x,y: x if x["marks"]>y["marks"] else y,passed_students)

print("students who passed:",passed_students)
print("topper among the passed students:",topper)
```

```python
#Question 15
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
#Using filter(), map(), and reduce():

#Extract students scoring more than 70.
#Extract only their marks.
#Calculate the average marks.

greater_than_70 = list(filter(lambda x: x["marks"]>70,students))
marks = list(map(lambda x: x["marks"],greater_than_70))
avg_marks = reduce(lambda x,y: x+y,marks)/len(marks)

print("students scoring more than 70:",greater_than_70)
print("marks:",marks)
print("average marks:",avg_marks)
```

```python
#Question 16
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
#Using filter(), map(), and reduce():

#Extract products currently in stock.
#Calculate each product's inventory value using price × stock.
#Calculate the total inventory value.

in_stock = list(filter(lambda x: x["stock"] !=0,products))
inventory_value = list(map(lambda x: x["price"]*x["stock"],in_stock))

total_inventory_value = reduce(lambda x,y: x+y,inventory_value)

print("products currently in stock:",in_stock)
print("inventory value:",inventory_value)
print("total inventory value:",total_inventory_value)
```

```python
#Question 17
products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]
#Using filter(), map(), and reduce():

#Extract products belonging to the Electronics category.
#Apply a 20% discount.
#Calculate the total discounted value.

electronics = list(filter(lambda x: x["category"]=="Electronics",products))
discount = list(map(lambda x: {"name":x["name"],"price":x["price"]*0.8},electronics))
total_discounted_value = reduce(lambda x,y: x+y["price"],discount,0)

print("products belonging to the Electronics category:",electronics)
print("discounted prices:",discount)
print("total discounted value:",total_discounted_value)
```

```python
#Question 19
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]
#Using filter(), map(), and reduce():

#Extract transactions whose amount is greater than 5000.
#Add 18% GST to every selected transaction.
#Calculate the grand total after GST.

greater_than_5000 = list(filter(lambda x: x["amount"]>5000,transactions))
add_gst = list(map(lambda x: {"id":x["id"],"amount":x["amount"]*1.18},greater_than_5000))
total_grand_total = reduce(lambda x,y: x+y["amount"],add_gst, 0)

print("transactions whose amount is greater than 5000:",greater_than_5000)
print("transactions after adding GST:",add_gst)
print("grand total after GST:",total_grand_total)
```

```python
#Question 20
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
#Using filter(), map(), and reduce():

#Extract active salespeople from the South region.
#Calculate a 5% incentive for each selected salesperson.
#Create a list containing salesperson name, sales amount, and incentive.
#Calculate the total incentive payable.

active_salespeople_south = list(filter(lambda x: x["active"]==True and x["region"]=="South",sales_records))
incentive_5 = list(map(lambda x: {"salesperson":x["salesperson"],"sales":x["sales"],"incentive":x["sales"]*0.05},active_salespeople_south))
total_incentive = reduce(lambda x,y: x+y["incentive"],incentive_5,0)

print("active salespeople from the South region:",active_salespeople_south)
print("incentive for each selected salesperson:",incentive_5)
print("total incentive payable:",total_incentive)









```
