```python
def my_decorator(func):
  ### it will add extra functionality
  ### around the original function
  def wrapper():
    print("Before the function")
    func()
    print("After the function")
  return wrapper
@my_decorator
def welcome():
  print("This is decorator example")
welcome()
```

```python
Original welcome()
        │
        ▼
my_decorator(welcome)
        │
        │ func = welcome
        ▼
    wrapper()
        │
        ├── print("Before")
        │
        ├── func()
        │      │
        │      └── Welcome to Python
        │
        └── print("After")
```

```python
# Decorator
def login_required(func):

    # Wrapper function
    def wrapper():

        # Assume user login status
        logged_in = False

        # Check login before executing
        # the original function.
        if logged_in:

            print("Login successful")

            # Execute original function
            func()

        else:

            print("Please login first")

    # Return modified/wrapped function
    return wrapper


# Apply decorator
@login_required
def employee_dashboard():

    print("Welcome to Employee Dashboard")


# Call function
employee_dashboard()
```

```python
def login_required(func):
    def wrapper(account, *args, **kwargs):
        if account["logged_in"]:
            print("Authentication successful")
            return func(account, *args, **kwargs)
        else:
            print("Authentication failed. Please login first.")
    return wrapper


def transaction_logger(func):
    def wrapper(account, *args, **kwargs):
        print("\nTransaction started")
        print("Account Holder:", account["name"])
        print("Account Number:", account["account_number"])

        result = func(account, *args, **kwargs)

        print("Transaction completed")
        print("Available Balance:", account["balance"])
        print("-" * 40)

        return result
    return wrapper


def sufficient_balance(func):
    def wrapper(account, amount, *args, **kwargs):
        if amount <= 0:
            print("Invalid transaction amount")
            return

        if account["balance"] >= amount:
            return func(account, amount, *args, **kwargs)
        else:
            print("Insufficient balance")
            print("Requested Amount:", amount)
            print("Available Balance:", account["balance"])
    return wrapper


@login_required
@transaction_logger
@sufficient_balance
def withdraw(account, amount):
    account["balance"] -= amount
    print("Withdrawn Amount:", amount)


@login_required
@transaction_logger
def deposit(account, amount):
    if amount <= 0:
        print("Invalid deposit amount")
        return

    account["balance"] += amount
    print("Deposited Amount:", amount)


@login_required
def check_balance(account):
    print("\nAccount Holder:", account["name"])
    print("Available Balance:", account["balance"])


account = {
    "name": "Bharath",
    "account_number": "ACC1001",
    "balance": 50000,
    "logged_in": True
}


print("BANKING APPLICATION")
print("=" * 40)

check_balance(account)

withdraw(account, 10000)

deposit(account, 5000)

withdraw(account, 60000)

check_balance(account)
```

```python
# Decorator to check whether customer is logged in
def login_required(func):

    # Wrapper receives product, quantity and customer information
    def wrapper(product, quantity, customer):

        # Check customer's login status
        if customer["logged_in"]:

            # Customer is authenticated
            print("Customer authentication successful")

            # Continue to the next decorator
            return func(product, quantity, customer)

        else:

            # Stop the purchase if customer is not logged in
            print("Please login before purchasing")

    # Return wrapper function
    return wrapper


# Decorator to check product stock
def check_stock(func):

    # Wrapper receives product, quantity and customer
    def wrapper(product, quantity, customer):

        # Check whether requested quantity is available
        if product["stock"] >= quantity:

            # Display stock availability
            print("Product is available")

            # Continue to the next decorator
            return func(product, quantity, customer)

        else:

            # Display error when requested quantity is unavailable
            print("Insufficient stock")

            # Display currently available quantity
            print("Available Stock:", product["stock"])

    # Return wrapper
    return wrapper


# Decorator to calculate customer discount
def apply_discount(func):

    # Wrapper receives product, quantity and customer
    def wrapper(product, quantity, customer):

        # Calculate original order amount
        original_amount = product["price"] * quantity

        # Check whether customer is a premium customer
        if customer["membership"] == "premium":

            # Premium customers receive 10% discount
            discount = original_amount * 0.10

        else:

            # Regular customers don't receive this discount
            discount = 0

        # Calculate amount after discount
        final_amount = original_amount - discount

        # Display original amount
        print("Original Amount:", original_amount)

        # Display discount amount
        print("Discount:", discount)

        # Display final payable amount
        print("Final Amount:", final_amount)

        # Call the actual purchase function
        return func(
            product,
            quantity,
            customer,
            final_amount
        )

    # Return wrapper
    return wrapper


# Decorator for logging order information
def order_logger(func):

    # Wrapper accepts any arguments
    def wrapper(*args, **kwargs):

        # Display message before order processing
        print("\nOrder processing started")

        # Execute next decorator/original function
        result = func(*args, **kwargs)

        # Display message after order processing
        print("Order processing completed")

        # Print separator
        print("-" * 50)

        # Return result
        return result

    # Return wrapper
    return wrapper


# Apply login authentication first
@login_required

# Check whether requested product quantity is available
@check_stock

# Calculate discount based on membership
@apply_discount

# Log order execution
@order_logger

# Actual business function responsible for purchasing product
def purchase_product(
    product,
    quantity,
    customer,
    final_amount
):

    # Reduce purchased quantity from inventory
    product["stock"] -= quantity

    # Display customer name
    print("Customer:", customer["name"])

    # Display purchased product
    print("Product:", product["name"])

    # Display purchased quantity
    print("Quantity:", quantity)

    # Display amount paid
    print("Amount Paid:", final_amount)

    # Display remaining inventory
    print("Remaining Stock:", product["stock"])

    # Display success message
    print("Purchase successful")


# Create product information
product = {

    # Product name
    "name": "Laptop",

    # Price of one laptop
    "price": 50000,

    # Available inventory
    "stock": 10
}


# Create customer information
customer = {

    # Customer name
    "name": "Bharath",

    # Customer login status
    "logged_in": True,

    # Customer membership type
    "membership": "premium"
}


# Purchase two laptops
purchase_product(
    product,
    2,
    customer
)
```

```python
class RegularCustomer:

    def calculate_discount(self, amount):

        return amount * 0.05


class PremiumCustomer:

    def calculate_discount(self, amount):

        return amount * 0.10


class VIPCustomer:

    def calculate_discount(self, amount):

        return amount * 0.20


customers = [
    RegularCustomer(),
    PremiumCustomer(),
    VIPCustomer()
]


amount = 10000


for customer in customers:

    discount = customer.calculate_discount(amount)

    print("Discount:", discount)
```

```python
class LinearRegression:

    def predict(self, data):
        print("Predicting using Linear Regression")
        return 25000


class RandomForest:

    def predict(self, data):
        print("Predicting using Random Forest")
        return 27500


class XGBoost:

    def predict(self, data):
        print("Predicting using XGBoost")
        return 29000


models = [
    LinearRegression(),
    RandomForest(),
    XGBoost()
]


for model in models:

    prediction = model.predict([10, 20, 30])

    print("Prediction:", prediction)
```
