```python
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [45000, 52000, 48000, 61000, 70000, 76000],
    "Profit": [8000, 9500, 8500, 12000, 15000, 17000],
    "Customers": [120, 140, 135, 165, 190, 210],
    "Ad_Spend": [5000, 6000, 5500, 7000, 8500, 9000]
}

df = pd.DataFrame(data)

print(df)
```
