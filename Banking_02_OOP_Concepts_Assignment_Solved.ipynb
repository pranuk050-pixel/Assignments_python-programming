### Q1. Create a account class for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# A bank account just needs to remember who owns it, what it holds, and stop
# the balance from going negative by accident.
class Account:
    def __init__(self, ref, holder_name, balance=0):
        self.ref = ref
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

acc = Account("ACC-1001", "Anita Rao", balance=5000)
acc.deposit(1500)
print(acc.ref, acc.holder_name, acc.balance)
# 6500 -> starting balance + the deposit. Nothing fancy yet, just state + one action.
```

### Q7. Compare two account references with __eq__ for banking case 1; use sample reference `ACC-1007` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Two Account objects should be considered "the same account" if their
# reference numbers match, even if they're different Python objects in memory.
class Account:
    def __init__(self, ref, balance=0):
        self.ref = ref
        self.balance = balance

    def __eq__(self, other):
        if not isinstance(other, Account):
            return NotImplemented
        return self.ref == other.ref

a1 = Account("ACC-1007", 1000)
a2 = Account("ACC-1007", 9999)   # different balance, same ref
a3 = Account("ACC-9999", 1000)
print(a1 == a2, a1 == a3)
```

### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# A property lets balance be read like a normal attribute but blocks anyone
# from setting it to something invalid, like a negative number.
class Account:
    def __init__(self, ref, balance=0):
        self.ref = ref
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("balance cannot go negative")
        self._balance = value

acc = Account("ACC-1008", 2000)
acc.balance = 2500          # fine
print(acc.balance)
try:
    acc.balance = -50
except ValueError as e:
    print("rejected:", e)
```

### Q9. Create portfolio and add_item for banking case 1; use sample reference `ACC-1009` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# A Portfolio groups several accounts under one customer view - add_account
# keeps that logic in one place instead of the caller managing a raw list.
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.accounts = []

    def add_account(self, account_ref, balance):
        self.accounts.append({"ref": account_ref, "balance": balance})

    def total_balance(self):
        return sum(a["balance"] for a in self.accounts)

p = Portfolio("Anita Rao")
p.add_account("ACC-1009", 15000)
p.add_account("ACC-2050", 8000)
print(p.owner, "total across accounts:", p.total_balance())
```

### Q12. Use a class method as an alternate constructor for banking case 1; use sample reference `ACC-1012` and explain the result.

2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Sometimes an account arrives as a single string, e.g. from a CSV row like
# "ACC-1012:5000". A classmethod gives a clean second way to build the object.
class Account:
    def __init__(self, ref, balance):
        self.ref = ref
        self.balance = balance

    @classmethod
    def from_csv_row(cls, row):
        ref, balance = row.split(":")
        return cls(ref, float(balance))

acc = Account.from_csv_row("ACC-1012:7400")
print(acc.ref, acc.balance)
```

### Q44. Implement a repository abstraction for banking case 1; use sample reference `ACC-1044` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# The repository hides *how* accounts are stored (a dict here, a database in
# production) behind one small interface: get/save. Callers never touch storage directly.
from abc import ABC, abstractmethod

class AccountRepository(ABC):
    @abstractmethod
    def get(self, ref): ...
    @abstractmethod
    def save(self, ref, balance): ...

class InMemoryAccountRepository(AccountRepository):
    def __init__(self):
        self._store = {}

    def get(self, ref):
        return self._store.get(ref)

    def save(self, ref, balance):
        self._store[ref] = balance

repo = InMemoryAccountRepository()
repo.save("ACC-1044", 18000)
print(repo.get("ACC-1044"))
```

### Q75. Implement a unit-of-work transaction for banking case 1; use sample reference `ACC-1075` and explain the result.

**Hints**

3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# A unit of work batches several account changes and only makes them
# permanent together - if anything fails partway, everything rolls back so
# the books never end up half-updated.
class UnitOfWork:
    def __init__(self, accounts):
        self.accounts = accounts
        self._snapshot = None

    def __enter__(self):
        self._snapshot = dict(self.accounts)   # remember original state
        return self

    def __exit__(self, exc_type, exc, tb):
        if exc_type is not None:
            self.accounts.clear()
            self.accounts.update(self._snapshot)   # rollback
            print("rolled back, nothing was committed")
            return True   # swallow the error, we've handled it
        print("committed:", self.accounts)

accounts = {"ACC-1075": 5000, "ACC-5002": 3000}
try:
    with UnitOfWork(accounts):
        accounts["ACC-1075"] -= 1000
        accounts["ACC-5002"] += 1000
        raise RuntimeError("downstream ledger service timed out")
except Exception:
    pass
print("final state:", accounts)
```

```python
# A unit of work batches several account changes and only makes them
# permanent together - if anything fails partway, everything rolls back so
# the books never end up half-updated.
class UnitOfWork:
    def __init__(self, accounts):
        self.accounts = accounts
        self._snapshot = None

    def __enter__(self):
        self._snapshot = dict(self.accounts)   # remember original state
        return self

    def __exit__(self, exc_type, exc, tb):
        if exc_type is not None:
            self.accounts.clear()
            self.accounts.update(self._snapshot)   # rollback
            print("rolled back, nothing was committed")
            return True   # swallow the error, we've handled it
        print("committed:", self.accounts)

accounts = {"ACC-1085": 5000, "ACC-5002": 3000}
try:
    with UnitOfWork(accounts):
        accounts["ACC-1085"] -= 1000
        accounts["ACC-5002"] += 1000
        raise RuntimeError("downstream ledger service timed out")
except Exception:
    pass
print("final state:", accounts)
```

### Q91. Design an extensible pricing engine for banking case 3; use sample reference `ACC-1091` and explain the result.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# The engine doesn't know about specific rate rules - it just runs whatever
# list of rules it's given. Adding a new rule later means adding a class,
# not editing the engine itself (open/closed principle).
class BaseRateRule:
    def apply(self, balance):
        return balance * 0.03

class SeniorCitizenRule:
    def apply(self, balance):
        return balance * 0.045

class HighBalanceRule:
    def apply(self, balance):
        return balance * 0.055 if balance > 100000 else balance * 0.03

class InterestPricingEngine:
    def __init__(self, rules):
        self.rules = rules

    def best_rate_return(self, balance):
        return max(rule.apply(balance) for rule in self.rules)

engine = InterestPricingEngine([BaseRateRule(), SeniorCitizenRule(), HighBalanceRule()])
print("account ACC-1091 best interest option:", round(engine.best_rate_return(150000), 2))
```

### Q92. Implement an observer for low balance for banking case 3; use sample reference `ACC-1092` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Account doesn't need to know who cares about low balances - it just
# announces the event. An SMS alert and a branch report both listen the same way.
class Account:
    def __init__(self, ref, balance, threshold=1000):
        self.ref, self.balance, self.threshold = ref, balance, threshold
        self._observers = []

    def subscribe(self, observer):
        self._observers.append(observer)

    def withdraw(self, amount):
        self.balance -= amount
        if self.balance < self.threshold:
            for obs in self._observers:
                obs(self.ref, self.balance)

def sms_alert(ref, balance):
    print(f"SMS: {ref} balance dropped to INR {balance}, please top up")

acc = Account("ACC-1092", 1200)
acc.subscribe(sms_alert)
acc.withdraw(500)
```
