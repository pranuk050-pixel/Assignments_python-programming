1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Matplotlib is ready!')
```

2. Load and Inspect Sample Data

```python
sales_data = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'sales': [120, 150, 170, 160, 210, 250]
})

print(sales_data)
print('\nShape:', sales_data.shape)
print('\nColumns:', sales_data.columns.tolist())
print('\nHead:')
print(sales_data.head())
print('\nDescription:')
print(sales_data.describe())
```

3. Create a Simple Line Plot

```python
plt.figure(figsize=(8, 5))
plt.plot(sales_data['month'], sales_data['sales'], color='blue', marker='o', linewidth=2)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Overall trend: Check if sales are generally increasing, decreasing, or stable across months. A steady upward slope signals growth momentum.

Seasonal peaks: Identify months with sharp increases — these could align with holidays, promotions, or seasonal demand.

Low-performing months: Spot dips where sales fall below average. These months may need targeted marketing or operational improvements.

Volatility: Large fluctuations between months suggest unstable demand or external influences.

Growth opportunities: Months showing consistent improvement can be leveraged for scaling campaigns or expanding inventory.

Comparative performance: Ranking months by sales helps prioritize resource allocation and budget planning.

4. Create a Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 55]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color=['#4c72b0', '#55a868', '#c44e52', '#8172b2'])
plt.title('Product Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
for i, v in enumerate(values):
    plt.text(i, v + 2, str(v), ha='center')
plt.tight_layout()
plt.show()
```

```python
# insights
# Category D leads: With 55 units, Category D is the strongest performer, significantly ahead of others.

# Category B strong: At 40 units, Category B shows solid demand, second only to D.

# Category C moderate: With 30 units, Category C sits in the mid-range, performing better than A but below B and D.

# Category A weakest: At 25 units, Category A lags behind, suggesting potential need for improvement or reevaluation.

# Sales distribution: There’s a clear gap between top (D) and bottom (A), highlighting uneven performance across categories.

# Growth focus: Categories B and C could be scaled further, while A may need marketing or product adjustments.
```

5. Create a Scatter Plot

```python
x = [1, 2, 3, 4, 5, 6, 7]
y = [2, 4, 5, 8, 10, 11, 15]

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='green', s=80, alpha=0.8)
plt.title('Study Hours vs Grades')
plt.xlabel('Hours Studied')
plt.ylabel('Score')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# insights :
# Positive correlation: As study hours increase, scores rise consistently, showing a clear link between effort and performance.

# Strong improvement zone: Beyond 4–5 hours, scores jump more sharply, suggesting extra study time yields higher returns.

# Top performance: Students studying 6–7 hours achieve the highest scores (11–15), indicating maximum benefit at higher commitment levels.

# Low study impact: With only 1–2 hours, scores remain low (2–4), reinforcing that minimal effort leads to weaker outcomes.

# Consistent upward trend: No major dips are visible — the relationship is steadily positive without anomalies.
```

6. Create a Histogram

```python
scores = [55, 60, 65, 66, 70, 72, 74, 75, 80, 82, 85, 90, 92, 95]

plt.figure(figsize=(8, 5))
plt.hist(scores, bins=5, color='purple', edgecolor='yellow')
plt.title('Distribution of Exam Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# insights
# Score concentration: Most students scored between 70–95, showing a strong cluster in the higher range.

# High achievers: A notable number of students achieved 85+, with several reaching 90–95, indicating excellent performance.

# Moderate performers: Scores in the 60–75 range are fairly common, representing average achievement.

# Low scores: Few students scored below 60, suggesting that underperformance is relatively rare.

# Distribution shape: The histogram skews toward higher scores, reflecting overall strong exam results.
```

7. Create a Pie Chart

```python
labels = ['A', 'B', 'C', 'D']
sizes = [30, 25, 20, 25]

plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#66c2a5', '#fc8d62', '#8da0cb', '#e78ac3'])
plt.title('Market Share Distribution')
plt.tight_layout()
plt.show()
```

```python
#INSIGHTS
# Category A largest share: Holds 30%, making it the leading segment in the market.

# Category B and D equal: Each at 25%, they share the second position, together accounting for half the market.

# Category C smallest: At 20%, it lags behind the others, suggesting weaker presence or lower demand.

# Balanced distribution: No single category dominates overwhelmingly; the shares are relatively close, indicating competitive dynamics.

# Growth potential: Categories B and D could be scaled to challenge A, while C may need strategic focus to improve its position.
```

9. Add Labels, Titles, and Legends

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_a = [20, 25, 23, 30, 35]
product_b = [15, 18, 22, 26, 31]

plt.figure(figsize=(8, 5))
plt.plot(months, product_a, marker='o', label='Product A')
plt.plot(months, product_b, marker='o', label='Product B', linestyle='--')
plt.title('Monthly Comparison')
plt.xlabel('Month')
plt.ylabel('Units Sold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# #insights
# Product A stronger: Consistently higher sales than Product B across all months, showing clear dominance.

# Product B catching up: Starts lower but shows steady growth, narrowing the gap by May.

# Steady upward trend: Both products increase month by month, indicating healthy demand growth.

# Gap analysis: The difference between A and B is largest early on but reduces later, suggesting B is improving faster.

# Peak performance: May records the highest units sold for both products, marking strong end-of-period performance.
```

Many Graph Examples Using the Seaborn Tips Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')
print(tips.head())
```

```python
avg_bill = tips.groupby('day')['total_bill'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.plot(avg_bill.index, avg_bill.values, marker='o', color='royalblue', linewidth=2)
plt.title('1. Line Plot: Average Bill by Day')
plt.xlabel('Day')
plt.ylabel('Average Total Bill')
plt.grid(True, alpha=0.4)
plt.show()
```

```python
# 📊 Key Observations
# Day-to-day variation: The plot highlights how spending patterns differ depending on the day of the week. Some days consistently show higher average bills, while others are lower.

# Weekend vs weekday: Typically, weekends (Saturday and Sunday) tend to have higher average bills compared to weekdays, reflecting larger groups dining out or more indulgent spending.

# Customer behavior: Peaks in the line plot suggest days when customers are more likely to order bigger meals or dine in larger groups.

# Operational insights: Restaurants could use this data to plan staffing, inventory, and promotions. For example, if Saturday shows the highest average bill, managers might allocate more staff or run premium offers
```

```python
# 2. Bar chart: average tip by day
avg_tip = tips.groupby('day')['tip'].mean().sort_index()
plt.figure(figsize=(8, 4))
plt.bar(avg_tip.index, avg_tip.values, color='green', edgecolor='black')
plt.title('2. Bar Chart: Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

```python
# 📊 Key Insights
# Day-to-day variation: Just like bills, tips vary by day. Some days consistently show higher average tips, while others are lower.

# Weekend generosity: Saturdays and Sundays often have higher average tips, which aligns with larger bills and more relaxed dining experiences.

# Weekday moderation: Tips on weekdays tend to be lower, possibly reflecting smaller groups or quicker meals.

# Correlation with bills: Since tips are usually a percentage of the bill, higher bills on weekends naturally lead to higher tips. But the bar chart helps confirm whether tipping behavior is proportional or if customers tip more generously on certain days.
```

```python
# 3. Bar chart: average bill by sex
sex_bill = tips.groupby('sex')['total_bill'].mean()
plt.figure(figsize=(7, 4))
plt.bar(sex_bill.index, sex_bill.values, color=['pink', 'skyblue'])
plt.title('3. Bar Chart: Average Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Total Bill')
plt.show()
```

```python
# 📊 Key Insights
# Average spending difference: One group (male or female) tends to have a higher average bill than the other. This suggests differences in dining behavior, group size, or ordering patterns.

# Group dynamics: If males show higher average bills, it could be linked to larger party sizes or more expensive menu choices. If females show higher averages, it may reflect different dining occasions or preferences.

# Restaurant strategy: Understanding which demographic spends more can help restaurants tailor promotions, menu items, or marketing campaigns.
```

```python
# 4. Histogram: total bill distribution
plt.figure(figsize=(8, 4))
plt.hist(tips['total_bill'], bins=15, color='purple', edgecolor='black')
plt.title('4. Histogram: Total Bill Distribution')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# 📊 Key Insights
# Central tendency: Most bills cluster in the $10–$20 range, meaning the majority of dining experiences fall within this moderate spending bracket.

# Spread: The distribution extends up to around $50, with fewer observations at the higher end. This indicates occasional large bills but they’re relatively rare.

# Skewness: The histogram is right-skewed (longer tail on the higher side). This suggests that while most customers spend moderately, a few spend significantly more.

# Frequency peaks: The tallest bars highlight the most common bill ranges, which can help restaurants identify their “sweet spot” for typical customer spending.
```

```python
# 5. Scatter plot: total bill vs tip
plt.figure(figsize=(8, 4))
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.7, color='darkorange')
plt.title('5. Scatter Plot: Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# 📊 Key Insights
# Positive correlation: As the total bill increases, the tip generally increases too. This makes sense since tips are often calculated as a percentage of the bill.

# Variation in tipping behavior: The points don’t form a perfect line — some customers tip more generously than others for the same bill size. This shows tipping isn’t strictly proportional.

# Outliers: A few points stand out where either:

# The bill is very high but the tip is relatively low (stingy tipping).

# The bill is moderate but the tip is unusually high (generous tipping).

# Tip percentage differences: The scatter plot suggests that while tips grow with bills, the percentage tipped may decline slightly for very large bills — customers might cap their tips at a certain level
```

```python
# 6. Scatter plot with colored sex groups
plt.figure(figsize=(8, 4))
for sex_group, group in tips.groupby('sex'):
    plt.scatter(group['total_bill'], group['tip'], label=sex_group, alpha=0.7)
plt.title('6. Scatter Plot: Bill vs Tip by Sex')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.legend()
plt.show()
```

```python
# 📊 Key Insights
# Positive correlation: Both male and female groups show the same general upward trend — higher bills tend to come with higher tips.

# Group differences: The spread of points may differ slightly:

# One group might cluster more tightly, suggesting consistent tipping behavior.

# The other may show wider variation, indicating more diverse tipping habits.

# Outliers: Outliers are visible in both groups — some very high bills with modest tips, and some moderate bills with unusually generous tips.

# Tip percentage variation: By coloring the groups, you can start to see whether one sex tends to tip proportionally more relative to the bill size.
```

```python
# 7. Box plot: total bill by day
plt.figure(figsize=(8, 4))
plt.boxplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
            labels=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('7. Box Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# 8. Box plot: total bill by smoker status
plt.figure(figsize=(7, 4))
plt.boxplot([tips[tips['smoker'] == s]['total_bill'] for s in ['Yes', 'No']],
            labels=['Smoker', 'Non-Smoker'])
plt.title('8. Box Plot: Total Bill by Smoker Status')
plt.ylabel('Total Bill')
plt.show()
```

```python
# 📊 Key Insights
# Median comparison: The median line shows whether smokers or non-smokers typically spend more. Often, non-smokers have slightly higher median bills, but this depends on the dataset sample.

# Spread of bills: The box height (IQR) indicates variability. If smokers show a wider spread, it suggests more diverse spending patterns within that group.

# Outliers: Both groups may have outliers — unusually high bills that represent large parties or special occasions.

# Behavioral differences: The plot helps identify whether smoking status correlates with consistent spending habits or more variability.
```

```python
# 9. Violin plot: total bill by day
plt.figure(figsize=(8, 4))
plt.violinplot([tips[tips['day'] == d]['total_bill'] for d in ['Thur', 'Fri', 'Sat', 'Sun']],
               showmeans=True)
plt.xticks([1,2,3,4], ['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('9. Violin Plot: Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# 📊 Key Insights
# Weekend spending: Saturday and Sunday show wider violins and higher means, indicating greater variability and higher average bills — typical of larger groups or more indulgent weekend dining.

# Weekday consistency: Thursday and Friday have narrower distributions, suggesting more predictable, moderate spending.

# Skewness and density: The bulges around $15–$20 show that most bills cluster in this range, while the long upper tails (especially on Saturday) reveal occasional high spenders.

# Mean vs median: The mean lines are slightly above the median for weekend days, confirming a right-skewed distribution — a few large bills pull the average upward.
```

```python
# 10. Count plot: number of observations by day
counts = tips['day'].value_counts().reindex(['Sun', 'Sat', 'Thur', 'Fri'])
plt.figure(figsize=(8, 4))
plt.bar(counts.index, counts.values, color='coral')
plt.title('10. Count Plot: Number of Observations by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

```python
# 📊 Key Insights
# Weekend dominance: Saturday has the highest count (~88), followed by Sunday (~75). This suggests weekends are the busiest days, with more customers dining out.

# Weekday drop: Thursday (~62) and Friday (~18) have fewer observations, indicating lighter traffic. Friday’s low count might reflect fewer recorded data points or genuinely lower customer turnout.

# Data distribution: The dataset is not evenly distributed — weekend data dominates. This imbalance should be considered when analyzing averages or correlations by day.
```

```python
# 11. Pie chart: smoker share
smoker_counts = tips['smoker'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(smoker_counts.values, labels=smoker_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('11. Pie Chart: Smoker Share')
plt.show()
```

```python
# 📊 Key Insights
# Majority non-smokers: About 61.9% of customers are non-smokers, forming the larger share of the dataset.

# Smoker minority: Roughly 38.1% are smokers, indicating a smaller but significant portion of the clientele.

# Behavioral implications: This split can influence spending and tipping patterns — smokers might have different dining habits or group sizes compared to non-smokers.

# Data balance: The dataset is moderately imbalanced, so analyses comparing smoker vs non-smoker behavior should consider sample size differences.
```

```python
# 12. Heatmap: correlation matrix
corr = tips[['total_bill', 'tip', 'size']].corr()
plt.figure(figsize=(6, 5))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.index)), corr.index)
plt.title('12. Heatmap: Correlation Matrix')
plt.show()
```

```python
# 📊 Key Insights
# Strong positive correlation between total_bill and tip: As bills increase, tips tend to rise — a natural pattern since tips are often proportional to the bill amount.

# Moderate correlation between total_bill and size: Larger groups generally spend more, but not perfectly — some small groups can still have high bills.

# Weak correlation between tip and size: Tip amounts don’t scale directly with group size; individuals may tip differently depending on service quality or bill splitting.

# Diagonal values are all 1.0, representing perfect self-correlation for each variable.
```
