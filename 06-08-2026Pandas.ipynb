````markdown
# Data Visualization Techniques in Exploratory Data Analysis (EDA)

Data Visualization helps us understand patterns, trends, relationships, distributions, and outliers in data before building machine learning models.

---

# 1. Histogram

## Definition

A Histogram is used to visualize the distribution of a continuous numerical variable.

It divides data into intervals called bins and shows the frequency of observations in each bin.

## Best Used For

- Age Distribution
- Salary Distribution
- Exam Marks Distribution
- Product Price Distribution

## Example

Student Marks:

```text
45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90
```

Histogram Groups:

```text
40-50  -> 2
50-60  -> 2
60-70  -> 3
70-80  -> 3
80-90  -> 2
```

## Python Example

```python
import matplotlib.pyplot as plt

plt.hist(df['Salary'], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

## Interpretation

- Bell Shape → Normal Distribution
- Left Skewed → More high values
- Right Skewed → More low values

---

# 2. Bar Graph

## Definition

A Bar Graph is used for discrete/categorical variables.

It compares values across categories.

## Best Used For

- Sales by Product
- Students by Department
- Employees by Designation

## Example

| Department | Employees |
|------------|-----------|
| HR | 20 |
| IT | 50 |
| Finance | 15 |
| Sales | 40 |

## Python Example

```python
import seaborn as sns

sns.barplot(
    x='Department',
    y='Employees',
    data=df
)
```

## Interpretation

Higher bar indicates higher count/value.

---

# 3. Line Graph

## Definition

Line Graph shows trends over time.

Data points are connected using lines.

## Best Used For

- Stock Prices
- Daily Sales
- Temperature Trends
- Revenue Growth

## Example

| Date | Sales |
|--------|-------|
| 01-Jan | 100 |
| 02-Jan | 120 |
| 03-Jan | 150 |

## Python Example

```python
plt.plot(
    df['Date'],
    df['Sales']
)

plt.title("Sales Trend")
plt.show()
```

## Interpretation

- Upward Trend
- Downward Trend
- Seasonal Patterns

---

# 4. Pie Chart

## Definition

Pie Charts show percentage contribution of categories.

Each slice represents a category.

## Best Used For

- Market Share
- Department Budget
- Product Category Contribution

## Example

| Product | Sales % |
|----------|----------|
| A | 40 |
| B | 30 |
| C | 20 |
| D | 10 |

## Python Example

```python
plt.pie(
    df['Sales'],
    labels=df['Product'],
    autopct='%1.1f%%'
)

plt.show()
```

## Interpretation

Shows proportion of each category.

## Limitation

Not suitable when many categories exist.

---

# 5. Heatmap

## Definition

Heatmap represents numerical values using colors.

Darker colors indicate higher values.

## Best Used For

- Correlation Analysis
- Missing Value Analysis
- Feature Relationships

## Example Correlation Matrix

| | Age | Salary | Experience |
|----|----|----|----|
| Age | 1.0 | 0.7 | 0.8 |
| Salary | 0.7 | 1.0 | 0.9 |
| Experience | 0.8 | 0.9 | 1.0 |

## Python Example

```python
import seaborn as sns

corr = df.corr()

sns.heatmap(
    corr,
    annot=True
)
```

## Interpretation

```text
+1 = Strong Positive Correlation
0  = No Correlation
-1 = Strong Negative Correlation
```

---

# 6. Choropleth Maps (Geographical Plots)

## Definition

A Choropleth Map visualizes geographical data using color intensity.

Regions are colored based on values.

## Best Used For

- State-wise Sales
- Country Population
- Covid Cases
- Election Results

## Example

| State | Sales |
|---------|--------|
| Karnataka | 1000 |
| Telangana | 800 |
| Tamil Nadu | 1200 |

## Python Example

```python
import plotly.express as px

fig = px.choropleth(
    df,
    locations='State',
    color='Sales'
)

fig.show()
```

## Interpretation

Darker regions indicate higher values.

---

# 7. Box Plot

## Definition

A Box Plot visualizes the distribution of a continuous variable and identifies outliers.

## Components

```text
Minimum
Q1 (25%)
Median (50%)
Q3 (75%)
Maximum
Outliers
```

## Example

Salary Data:

```text
10, 15, 18, 20, 25, 28, 30, 35, 200
```

200 is an outlier.

## Python Example

```python
sns.boxplot(
    y=df['Salary']
)
```

## Interpretation

Useful for:

- Outlier Detection
- Skewness Detection
- Data Spread Analysis

---

# 8. Linear Model Plot

## Definition

Linear Model Plot shows the relationship between variables assuming a straight-line relationship.

Equation:

```text
Y = MX + C
```

## Best Used For

- Salary vs Experience
- Sales vs Advertisement Spend

## Example

| Experience | Salary |
|------------|---------|
| 1 | 30000 |
| 2 | 35000 |
| 3 | 40000 |

## Python Example

```python
sns.regplot(
    x='Experience',
    y='Salary',
    data=df
)
```

## Interpretation

Straight line indicates linear relationship.

---

# 9. Non-Linear Plot

## Definition

Shows curved relationships between variables.

Not all real-world data follows a straight line.

## Example

Population Growth:

```text
Y = X²
```

| X | Y |
|---|---|
| 1 | 1 |
| 2 | 4 |
| 3 | 9 |
| 4 | 16 |

## Python Example

```python
plt.scatter(x, y)

plt.plot(x, y)
```

## Interpretation

Useful when relationships are:

- Quadratic
- Exponential
- Logarithmic

---

# 10. Scatter Plot

## Definition

Scatter Plot displays relationships between two continuous variables.

Each point represents one observation.

## Best Used For

- Age vs Salary
- Height vs Weight
- Experience vs Income

## Example

| Age | Salary |
|------|--------|
| 25 | 30000 |
| 30 | 45000 |
| 35 | 60000 |

## Python Example

```python
plt.scatter(
    df['Age'],
    df['Salary']
)
```

## Interpretation

Positive Correlation:

```text
/
/
/
```

Negative Correlation:

```text
\
 \
  \
```

No Correlation:

```text
Random Points
```

---

# 11. Surface Plot (3D Plot)

## Definition

Surface Plot visualizes three-dimensional data.

Variables:

```text
X-axis
Y-axis
Z-axis
```

## Best Used For

- Scientific Simulations
- Machine Learning Cost Functions
- Weather Analysis

## Example

```text
Temperature
Latitude
Longitude
```

## Python Example

```python
from mpl_toolkits.mplot3d import Axes3D

ax.plot_surface(X, Y, Z)
```

## Interpretation

Shows peaks and valleys in data.

---

# 12. Sunburst Chart

## Definition

Sunburst Chart is used for hierarchical deep-dive analysis.

Shows parent-child relationships.

## Structure

```text
Company
│
├── Sales
│   ├── Product A
│   └── Product B
│
└── Marketing
    ├── Digital
    └── Offline
```

## Best Used For

- Organization Structure
- Product Hierarchy
- Sales Analysis
- Customer Segmentation

## Python Example

```python
import plotly.express as px

fig = px.sunburst(
    df,
    path=['Department','Team'],
    values='Revenue'
)

fig.show()
```

## Interpretation

Inner Circle:
Parent Category

Outer Circle:
Sub Category

Larger Segment:
Higher Contribution

---

# Visualization Selection Cheat Sheet

| Visualization | Variable Type | Purpose |
|--------------|--------------|----------|
| Histogram | Continuous | Distribution |
| Bar Graph | Categorical | Comparison |
| Line Graph | Date/Time | Trend Analysis |
| Pie Chart | Categorical | Percentage Share |
| Heatmap | Continuous | Correlation |
| Choropleth Map | Geographical | Regional Analysis |
| Box Plot | Continuous | Outlier Detection |
| Linear Plot | Continuous | Linear Relationship |
| Non-Linear Plot | Continuous | Curved Relationship |
| Scatter Plot | Continuous | Correlation Analysis |
| Surface Plot | 3 Dimensions | 3D Visualization |
| Sunburst Chart | Hierarchical | Deep Dive Analysis |

# EDA Visualization Workflow

```text
Dataset
   │
   ▼
Understand Data Types
   │
   ├── Continuous Variables
   │      ├── Histogram
   │      ├── Box Plot
   │      ├── Scatter Plot
   │      └── Heatmap
   │
   ├── Categorical Variables
   │      ├── Bar Chart
   │      └── Pie Chart
   │
   ├── Time Series Data
   │      └── Line Chart
   │
   ├── Geographic Data
   │      └── Choropleth Map
   │
   └── Hierarchical Data
          └── Sunburst Chart
```

These are the most commonly used visualization techniques in Data Science, Machine Learning, Business Intelligence, and Exploratory Data Analysis (EDA).
````
