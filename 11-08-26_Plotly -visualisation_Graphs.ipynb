Plotly

1. Import Required Libraries

```python
import pandas as pd
import plotly.express as px

print('Plotly is ready!')
```

2. Load and Inspect Sample Data

```python
#We will create a simple dataset and inspect it before creating charts.

sales_df = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'sales': [120, 150, 170, 165, 210, 245],
    'product': ['A', 'A', 'A', 'B', 'B', 'B']
})

print(sales_df)
print('\nShape:', sales_df.shape)
print('\nHead:')
print(sales_df.head())
print('\nSummary:')
print(sales_df.describe())
```

3. Create a Line Plot

```python
fig = px.line(
    sales_df,
    x='month',
    y='sales',
    title='Monthly Sales Trend',
    markers=True
)
fig.show()
```

```python
# 📊 Key Insights
# Overall trend: The line reveals whether sales are increasing, decreasing, or fluctuating month to month. A steady upward slope indicates growth momentum, while dips may signal seasonal slowdowns or external factors.

# Seasonal peaks: Months with sharp rises could correspond to promotional periods, holidays, or product launches. Identifying these helps plan future campaigns.

# Low-performing months: Troughs in the line highlight periods needing attention — perhaps due to reduced demand or marketing gaps.

# Volatility: The steepness of changes between months shows how stable or erratic sales are. Smooth curves suggest consistent performance; jagged ones imply variability.
```

```python
#4. Create a Bar Chart
#Bar charts are useful for comparing values across categories.

fig = px.bar(
    sales_df,
    x='month',
    y='sales',
    title='Monthly Sales by Month',
    color='month'
)
fig.show()
```

```python
# 📊 Key Insights
# Clear comparison: Unlike the line chart, the bar chart emphasizes differences between months, making it easy to spot which months outperform others.

# Top-performing months: The tallest bars highlight peak sales periods — these could align with seasonal demand, promotions, or product launches.

# Low-performing months: Shorter bars show weaker months, useful for identifying areas needing improvement.

# Seasonality: If certain months consistently stand out, this suggests cyclical demand patterns that can guide inventory and marketing strategies.

# Color encoding: Using month-based colors makes each category distinct, improving readability and quick comparisons.
```

```python
#5. Create a Scatter Plot
#Scatter plots show the relationship between two numerical variables.

scatter_data = pd.DataFrame({
    'hours_studied': [1, 2, 3, 4, 5, 6, 7],
    'score': [40, 50, 55, 65, 75, 82, 90]
})

fig = px.scatter(
    scatter_data,
    x='hours_studied',
    y='score',
    title='Study Hours vs Score',
    size_max=15
)
fig.show()
```

```python
# #📊 Key Insights
# Positive correlation: As study hours increase, scores rise. The upward trend is clear, showing that more time invested in studying generally leads to better results.

# Diminishing returns: The slope is steeper at lower hours (big gains from small increases), but flattens slightly at higher hours — suggesting that beyond ~6–7 hours, extra studying yields smaller improvements.

# Predictability: The points form a near-linear pattern, meaning scores can be fairly well predicted from study hours.

# Performance range: Scores range from ~40 (1 hour) to ~90 (7 hours), showing a strong payoff for consistent effort.
```

```python
#6. Create a Histogram
#A histogram shows the distribution of values in a single numeric variable.

scores = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 90, 92]

fig = px.histogram(
    x=scores,
    nbins=6,
    title='Distribution of Scores'
)
fig.update_layout(xaxis_title='Score', yaxis_title='Frequency')
fig.show()
```

```python
# 📊 Key Insights
# Central clustering: Most scores fall between 70–85, showing that the majority of students are performing in the upper-middle range.

# High achievers: A few scores extend into the 90+ range, highlighting top performers.

# Lower scores: Only a handful of scores are below 65, suggesting fewer struggling students.

# Distribution shape: The histogram is slightly right-skewed, with more students scoring higher than lower. This indicates overall strong performance.

# Frequency peaks: The tallest bins around 70–80 show the most common score range — the “sweet spot” of student achievement.
```

```python
#7. Create a Pie Chart
#Pie charts show the share of each category in a total.

labels = ['Product A', 'Product B', 'Product C', 'Product D']
values = [25, 30, 20, 25]

fig = px.pie(
    names=labels,
    values=values,
    title='Product Sales Share'
)
fig.show()
```

```python
# 📊 Key Insights
# Balanced distribution: Products A and D each hold 25%, showing equal contribution.

# Top performer: Product B leads with 30%, making it the largest contributor to total sales.

# Lowest share: Product C has 20%, slightly behind the others, suggesting room for improvement or targeted marketing.

# Portfolio balance: The shares are relatively close, indicating no single product dominates excessively — a healthy spread across the portfolio.
```

8. Create an Interactive Dashboard-Style Chart
Plotly is especially strong for interactive and dashboard-like visuals.

```python
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
product_a = [20, 25, 22, 30, 35, 40]
product_b = [15, 18, 20, 26, 28, 32]

fig = px.line(
    pd.DataFrame({
        'month': months,
        'Product A': product_a,
        'Product B': product_b
    }),
    x='month',
    y=['Product A', 'Product B'],
    title='Interactive Multi-Series Plot'
)
fig.update_layout(legend_title_text='Products')
fig.show()
```

```python
# 📊 Key Insights
# Growth trajectory: Both products show steady upward trends, indicating consistent sales growth across the months.

# Product A dominance: Product A starts higher and maintains a lead throughout, ending at 40 units in June compared to Product B’s 32.

# Product B catching up: While Product B lags behind, its growth rate is similar — suggesting potential to close the gap if momentum continues.

# Seasonal acceleration: Both products show sharper increases from April onward, possibly tied to seasonal demand or promotions.

# Parallel trends: The lines move in sync, implying external factors (like market demand or campaigns) affect both products similarly.
```

9. Add Custom Styling and Layout
Plotly offers many styling options such as colors, titles, and axis formatting.

```python
fig = px.line(
    sales_df,
    x='month',
    y='sales',
    title='Styled Plotly Chart',
    color_discrete_sequence=['#2E8B57']
)
fig.update_traces(mode='lines+markers', line=dict(width=3), marker=dict(size=9))
fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Sales',
    template='plotly_white'
)
fig.show()
```

```python
# 📊 Key Insights
# Clear upward trend: The line with markers shows whether sales are steadily increasing, plateauing, or fluctuating. If the slope is positive, it indicates growth momentum.

# Marker emphasis: Adding markers highlights each month’s exact value, making it easier to spot peaks and troughs.

# Strong visual clarity: The thicker line and larger markers improve readability, especially for presentations or dashboards.

# Seasonal patterns: Any sharp rises or dips across months may reflect seasonal demand, promotions, or external factors.

# Template choice: Using the plotly_white template ensures a clean, professional look that emphasizes the data without distractions.
```

10. Save and Final Notes

You can export Plotly figures as HTML files for sharing.

```python
fig = px.bar(
    sales_df,
    x='month',
    y='sales',
    title='Sales Chart for Export'
)
fig.write_html('plotly_sales.html')
print('Saved as plotly_sales.html')
fig.show()
```

```python
# 📊 Key Insights
# Month-to-month comparison: The bar chart highlights differences more starkly than a line chart, making it easy to spot which months outperform others.

# Top-performing months: The tallest bars show peak sales periods — these are critical for identifying seasonal demand or successful campaigns.

# Low-performing months: Shorter bars reveal weaker months, useful for diagnosing dips in performance.

# Trend visibility: While the chart doesn’t show continuity like a line chart, the gradual increase or decrease in bar heights still conveys the overall trend.

# Export utility: Saving as HTML makes the chart interactive and shareable, ensuring stakeholders can explore the data directly.
```

Many Graph Examples Using the Seaborn Tips Dataset

```python
import seaborn as sns
import plotly.express as px
```

```python
# Load dataset
tips = sns.load_dataset('tips')
print(tips.head())
```

```python
# 1. Line chart: average bill by day
avg_bill = tips.groupby('day', as_index=False)['total_bill'].mean()
fig = px.line(avg_bill, x='day', y='total_bill', title='1. Line Chart: Average Bill by Day', markers=True)
fig.show()
```

```python
# 📊 Key Insights
# Weekend boost: Saturday and Sunday typically show higher average bills, reflecting larger groups or more indulgent weekend dining.

# Weekday moderation: Thursday and Friday averages are lower, suggesting more consistent, moderate spending patterns.

# Day-to-day variation: The chart makes it easy to spot which day has the highest or lowest average — useful for operational planning.

# Customer behavior: The upward slope toward weekends indicates that dining habits shift, with customers spending more as leisure time increases.
```

```python
# 2. Bar chart: average tip by day
avg_tip = tips.groupby('day', as_index=False)['tip'].mean()
fig = px.bar(avg_tip, x='day', y='tip', title='2. Bar Chart: Average Tip by Day')
fig.show()
```

```python
# 📊 Key Insights
# Weekend tipping: Saturday and Sunday often show higher average tips, reflecting larger bills or more generous weekend diners.

# Weekday moderation: Thursday and Friday averages are typically lower, suggesting more consistent but modest tipping.

# Day-to-day variation: The chart makes it easy to spot which day has the highest or lowest average tip — useful for understanding customer generosity patterns.

# Correlation with bills: Since tips usually scale with bills, the pattern here likely mirrors your earlier average bill chart, but not perfectly — some days may show higher tipping relative to bill size.
```

```python
# 3. Scatter plot: total bill vs tip
fig = px.scatter(tips, x='total_bill', y='tip', color='sex', title='3. Scatter Plot: Total Bill vs Tip')
fig.show()
```

```python
📊 Key Insights
Positive correlation: As total bills increase, tips generally rise. This makes sense since tips are often proportional to the bill amount.

Gender differences: Coloring by sex lets you see whether male and female customers show different tipping patterns. Often, both follow the same upward trend, but one group may cluster slightly higher or lower.

Outliers: Some points show very high bills with relatively modest tips, or unusually generous tips on smaller bills. These outliers can skew averages and are worth investigating.

Tip efficiency: While the scatter shows raw amounts, adding a tip percentage line would reveal whether generosity scales consistently or declines as bills grow larger.
```

```python
# 4. Histogram: total bill distribution
fig = px.histogram(tips, x='total_bill', nbins=20, title='4. Histogram: Total Bill Distribution')
fig.show()
```

```python
# 📊 Key Insights
# Central clustering: Most bills fall between 10–20 dollars, showing that typical dining costs are modest.

# Long right tail: A few very high bills (40–50+) stretch the distribution, creating a right-skew. These represent larger groups or premium orders.

# Peak frequency: The tallest bins around 15–20 indicate the most common bill size — the “sweet spot” of customer spending.

# Outliers: Rare but notable bills above 50 stand out, worth analyzing separately to see if they correspond to special occasions or large parties.

# Distribution shape: The histogram is unimodal with a skew, meaning most customers spend within a narrow band, but a minority spend much more.
```

```python
# 5. Box plot: total bill by day
fig = px.box(tips, x='day', y='total_bill', title='5. Box Plot: Total Bill by Day')
fig.show()
```

```python
# 📊 Key Insights
# Median spending: The line inside each box shows the typical bill for that day. Weekends (Sat, Sun) usually have higher medians compared to weekdays.

# Spread of bills: The height of each box indicates variability. Larger spreads suggest more diverse spending behavior (e.g., bigger groups or varied orders).

# Outliers: Dots outside the whiskers highlight unusually high bills, often linked to large parties or special occasions.

# Day-to-day comparison: Thursday and Friday tend to have tighter ranges, while weekends show broader distributions, reflecting more variability in customer behavior.
```

```python
# 6. Bar chart: average bill by sex
sex_bill = tips.groupby('sex', as_index=False)['total_bill'].mean()
fig = px.bar(sex_bill, x='sex', y='total_bill', title='6. Bar Chart: Average Bill by Sex')
fig.show()
```

```python
# 📊 Key Insights
# Average spending comparison: One gender shows a slightly higher average bill than the other, suggesting differences in group size, dining choices, or spending habits.

# Consistency: The gap isn’t extreme, meaning both groups spend within a similar range overall.

# Behavioral patterns: If males tend to have higher averages, it could reflect larger parties or more expensive orders; if females lead, it may indicate different dining preferences.

# Operational relevance: Understanding these differences can help restaurants tailor promotions or menu options to customer segments.
```

```python
# 7. Scatter plot: bill vs party size
fig = px.scatter(tips, x='total_bill', y='size', color='day', title='7. Scatter Plot: Bill vs Party Size')
fig.show()
```

```python
# 📊 Key Insights
# Positive relationship: Larger parties generally have higher total bills, which makes sense since more people order more food.

# Day-based variation: Coloring by day shows whether certain days (like weekends) tend to have larger parties and higher bills. For example, Saturdays and Sundays often cluster at higher values.

# Outliers: Some small parties (size 2–3) have unusually high bills, suggesting premium orders or special occasions. Conversely, some larger groups have moderate bills, perhaps due to lighter dining.

# Distribution spread: Most points cluster around party sizes of 2–4, indicating typical dining groups, while larger sizes are less frequent but drive higher totals.
```

```python
# 8. Violin plot: bill by smoker status
fig = px.violin(tips, x='smoker', y='total_bill', box=True, title='8. Violin Plot: Total Bill by Smoker Status')
fig.show()
```

```python
# 📊 Key Insights
# Median comparison: The central line in each violin shows the median bill. If smokers and non-smokers differ, it highlights spending tendencies between the two groups.

# Distribution shape: The width of the violin at different bill values shows how common those amounts are. Wider sections mean more frequent bills in that range.

# Variability: Smokers may show a broader spread, suggesting more diverse spending behavior, while non-smokers might cluster more tightly.

# Outliers: The box overlay highlights unusually high bills, often linked to large parties or premium dining choices.

# Behavioral patterns: Differences in distribution could reflect lifestyle or group dynamics — for example, smokers dining in larger groups or ordering differently.
```

```python
# 9. Pie chart: day share
fig = px.pie(tips, names='day', title='9. Pie Chart: Day Share')
fig.show()
```

```python
# 📊 Key Insights
# Dominant days: If Saturday and Sunday slices are larger, it confirms weekends attract the majority of customers.

# Weekday contribution: Smaller slices for Thursday or Friday suggest fewer visits, but they still contribute meaningfully to overall traffic.

# Balanced vs skewed: If the chart looks balanced, customer flow is steady across days. If skewed, it highlights peak days that drive most of the business.

# Operational impact: Larger weekend shares imply higher staffing and inventory needs, while weekday shares show opportunities for promotions to boost traffic.

# Customer behavior: The distribution reflects dining habits — leisure-driven weekends vs routine-driven weekdays.
```

```python
# 10. Heatmap: correlation matrix
corr = tips[['total_bill', 'tip', 'size']].corr().round(2)
fig = px.imshow(corr, text_auto=True, color_continuous_scale='RdBu', title='10. Heatmap: Correlation Matrix')
fig.show()
```

```python
# 📊 Key Insights
# Total bill vs tip: Strong positive correlation — higher bills usually lead to higher tips. This is expected since tips are often proportional to the bill.

# Total bill vs party size: Moderate positive correlation — larger groups tend to generate bigger bills, though not perfectly (some small groups spend heavily).

# Tip vs party size: Weaker correlation compared to bill vs tip. Larger groups don’t always tip proportionally more, which could reflect differences in tipping behavior across group sizes.

# Matrix clarity: The color scale makes it easy to spot strong vs weak relationships — red for positive, blue for negative, and near-white for neutral.
```

```python
# 11. Line chart: tip by day
fig = px.line(tips.groupby('day', as_index=False)['tip'].mean(), x='day', y='tip', title='11. Line Chart: Average Tip by Day', markers=True)
fig.show()
```

```python
# 📊 Key Insights
# Weekend boost: Saturday and Sunday often show higher average tips, reflecting larger bills or more relaxed weekend dining behavior.

# Weekday moderation: Thursday and Friday averages are typically lower, suggesting steady but modest tipping during routine dining.

# Day-to-day variation: The chart makes it easy to spot which day has the highest or lowest tipping — useful for staff expectations and service planning.

# Correlation with bills: Since tips usually scale with bills, the pattern here likely mirrors the average bill chart, but not perfectly — some days may show higher generosity relative to bill size.

# Marker clarity: Adding markers emphasizes exact values, making peaks and troughs more visible.
```

```python
# 12. Bar chart: total bill by time
fig = px.bar(tips.groupby('time', as_index=False)['total_bill'].mean(), x='time', y='total_bill', title='12. Bar Chart: Average Bill by Meal Time')
fig.show()
```

```python
# 📊 Key Insights
# Dinner dominance: Dinner usually shows a higher average bill than lunch, reflecting larger groups, more courses, or premium menu choices.

# Lunch moderation: Lunch averages are typically lower, suggesting quicker meals, smaller parties, or lighter spending.

# Customer behavior: The contrast highlights how dining habits shift — lunch is functional, while dinner is more social and indulgent.

# Operational impact: Higher dinner bills mean restaurants should allocate more staff and inventory for evenings, while lunch may focus on efficiency and turnover.

# Revenue strategy: Promotions at lunch could raise average bills, while dinner menus can emphasize premium offerings to maximize already higher spending.
```
