### Q1. Write a function for line total for banking case 1; use sample reference `ACC-1001` and explain the result.

```python
def instalment_value(amount: float, count: int = 1) -> float:
    """Total value of `count` equal instalments of `amount`, rounded to paise."""
    return round(amount * count, 2)

# ACC-1001 — Rahul Verma: recurring-deposit instalments
instalments = [2750, 3250, 2500]
print(instalment_value(instalments[0], 4))          # 4 instalments of the first amount
print(list(map(lambda x: round(x * 1.05, 2), instalments)))   # with 5% top-up interest
print(list(filter(lambda x: x >= 2750, instalments)))       # instalments above the base
print(reduce(lambda a, b: a + b, instalments, 0))             # total value in this batch
```

**Explanation:** `instalment_value` only does one thing and doesn't touch anything outside itself, so it's easy to reuse for any account, not just ACC-1001. The `map`/`filter`/`reduce` lines below it are kept separate on purpose — one transforms, one selects, one totals — instead of cramming all three jobs into a single loop.

### Q4. Accept keyword arguments for a account for banking case 1; use sample reference `ACC-1004` and explain the result.

```python
def open_account(*, acc_no: str, holder: str, balance: float = 0.0, acc_type: str = "savings") -> dict:
    """Create a new account record. Keyword-only args stop callers from
    accidentally swapping holder name and account type by position."""
    return {"acc_no": acc_no, "holder": holder, "balance": balance, "type": acc_type}

new_acc = open_account(acc_no="ACC-1004", holder="Divya Iyer", balance=6000)
print(new_acc)
```

### Q6. Write a account docstring for banking case 1; use sample reference `ACC-1006` and explain the result.

```python
def calculate_interest(principal: float, annual_rate: float, years: float) -> float:
    """Calculate simple interest earned on a savings balance.

    Args:
        principal: opening balance in rupees.
        annual_rate: yearly interest rate as a decimal, e.g. 0.05 for 5%.
        years: how long the money stays deposited.

    Returns:
        The interest earned, rounded to 2 decimal places.

    Example:
        >>> calculate_interest(10000, 0.05, 2)
        1000.0
    """
    return round(principal * annual_rate * years, 2)

print(calculate_interest(10000, 0.05, 2))
print(calculate_interest.__doc__.splitlines()[0])   # confirm the docstring is attached
```

**Explanation:** A Google-style docstring like this shows up in `help()`, in IDE tooltips, and in generated docs, so anyone reusing the function later (including future-you) knows exactly what units it expects without opening the source.

### Q31. Compose discount and tax functions for banking case 1; use sample reference `ACC-1031` and explain the result.

```python
def apply_discount(amount, rate=0.07):
    return round(amount * (1 - rate), 2)

def apply_gst(amount, rate=0.18):
    return round(amount * (1 + rate), 2)

def compose(*funcs):
    """Chain functions right-to-left: compose(f, g)(x) == f(g(x))."""
    def combined(x):
        result = x
        for f in reversed(funcs):
            result = f(x if f is funcs[-1] else result)
        return result
    return combined

net_fee = compose(apply_gst, apply_discount)
print(net_fee(1300))   # ACC-1031: discount first, then GST on what's left
```

**Explanation:** `compose` lets two small, independently-testable functions (`apply_discount`, `apply_gst`) be chained without writing a third function that just calls both — useful once a pipeline grows past two or three steps.

### Q33. Group accounts using a helper function for banking case 1; use sample reference `ACC-1033` and explain the result.

```python
def group_by(items, key_func):
    """Group a list of dicts into {key: [items]} using key_func."""
    groups = {}
    for item in items:
        groups.setdefault(key_func(item), []).append(item)
    return groups

accounts = [
    {"acc_no": "ACC-1033", "type": "savings"},
    {"acc_no": "ACC-3011", "type": "current"},
    {"acc_no": "ACC-3012", "type": "savings"},
]
by_type = group_by(accounts, lambda a: a["type"])
print({k: [a["acc_no"] for a in v] for k, v in by_type.items()})
```

**Explanation:** `group_by` is written once and takes the grouping rule (`key_func`) as an argument, so the exact same helper works for grouping by account type today and by branch or risk tier tomorrow, with no code duplication.

### Q34. Create a closure for category discounts for banking case 1; use sample reference `ACC-1034` and explain the result.

```python
def interest_rate_for(tier):
    """Return a function that applies the interest rate for a given tier."""
    rates = {"silver": 0.04, "gold": 0.06, "platinum": 0.08}
    rate = rates.get(tier, 0.04)
    def apply(principal):
        return round(principal * (1 + rate), 2)
    return apply

gold_rate = interest_rate_for("gold")
print(gold_rate(7000))   # ACC-1034 is a gold-tier account
```

**Explanation:** `apply` remembers `rate` from `interest_rate_for` even after that outer function has returned — that's the closure. It means `gold_rate` can be handed around and called later without re-passing the tier every time.

### Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference `ACC-1035` and explain the result.

```python
from functools import wraps

def audit(label):
    def deco(fn):
        @wraps(fn)
        def wrapper(*args, **kwargs):
            print(f"[{label}] calling {fn.__name__} with {args}")
            return fn(*args, **kwargs)
        return wrapper
    return deco

@audit("TRANSFER")
def transfer_funds(from_acc, to_acc, amount):
    return f"Moved {amount} from {from_acc} to {to_acc}"

print(transfer_funds("ACC-1035", "ACC-4019", 900))
```

**Explanation:** `@wraps(fn)` keeps the wrapped function's real name and docstring intact, so `transfer_funds.__name__` still says `transfer_funds` instead of `wrapper` — small detail, but it matters once other tools (or `help()`) start inspecting the function.

### Q36. Build a map-filter-reduce transactions pipeline for banking case 1; use sample reference `ACC-1036` and explain the result.

```python
from functools import reduce
transactions = [
    {"acc_no": "ACC-1036", "amount": 2500, "status": "cleared"},
    {"acc_no": "ACC-1036", "amount": 1050, "status": "pending"},
    {"acc_no": "ACC-1036", "amount": 1500, "status": "cleared"},
]
cleared = filter(lambda t: t["status"] == "cleared", transactions)
amounts = map(lambda t: t["amount"], cleared)
total_cleared = reduce(lambda a, b: a + b, amounts, 0)
print("cleared total for", "ACC-1036", "=", total_cleared)
```

### Q78. Implement retry as a higher-transfer function for banking case 1; use sample reference `ACC-1078` and explain the result.

```python
import random

def retry(fn, times=3):
    """Call fn up to `times` times, returning the first successful result."""
    def wrapper(*args, **kwargs):
        last_error = None
        for attempt in range(1, times + 1):
            try:
                return fn(*args, **kwargs)
            except ConnectionError as e:
                last_error = e
                print(f"attempt {attempt} failed, retrying...")
        raise last_error
    return wrapper

random.seed(3)   # deterministic for this demo
def flaky_transfer(acc_no, amount):
    if random.random() < 0.5:
        raise ConnectionError("core banking timeout")
    return f"transferred {amount} from {acc_no}"

safe_transfer = retry(flaky_transfer, times=3)
print(safe_transfer("ACC-1078", 850))
```

**Explanation:** `retry` wraps any function that might fail with a transient error and reruns it a few times before giving up — the caller still just calls `safe_transfer(...)` and doesn't need to know retrying is happening underneath.

### Q81. Design a configurable transformation pipeline for banking case 2; use sample reference `ACC-1081` and explain the result.

```python
from functools import reduce

def build_pipeline(*steps):
    """Return one function that runs `steps` in order, left to right."""
    def run(data):
        return reduce(lambda acc_data, step: step(acc_data), steps, data)
    return run

transactions = [1900, 1100, 2400, 400]
pipeline = build_pipeline(
    lambda txs: filter(lambda x: x > 500, txs),
    lambda txs: map(lambda x: round(x * 1.18, 2), txs),
    lambda txs: list(txs),
)
print(pipeline(transactions))   # ACC-1081: filtered + GST-adjusted transactions
```

**Explanation:** The pipeline is just a list of functions run one after another with `reduce` — reordering steps, or adding a new one, means editing the `build_pipeline(...)` call and nothing else.
